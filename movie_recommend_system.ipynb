```python
import ast
```

```python
#Helper function
def converter(obj):
  L=[]
  for i in ast.literal_eval(obj):
    L.append(i['name'])
  return L
```

```python
movies['genres']=movies['genres'].apply(converter)
```

```python
movies.head()
```

```python
movies['keywords']=movies['keywords'].apply(converter)
```

```python
movies.head()
```

```python
def converter3(obj):
  c=0
  L=[]
  for i in ast.literal_eval(obj):
    if c<3:
      L.append(i['name'])
      c+=1
    else:
      return L
```

```python
movies['cast']=movies['cast'].apply(converter3)
```

```python
movies.head()
```

```python
movies['crew'].values
```

```python
def fetch_director(obj):
  L=[]
  for i in ast.literal_eval(obj):
    if i['job']=='Director':
      L.append(i['name'])
      break
  return L
```

```python
movies['crew']=movies['crew'].apply(fetch_director)
```

```python
movies.head()
```

```python
movies['overview']=movies['overview'].apply(lambda x:x.split())
```

```python
movies.head()
```

we have to delete space btw words because model understand it two different words example:sam Mendes to samMendes

```python
movies['genres']=movies['genres'].apply(lambda x:[i.replace(" ","") for i in x] if x is not None else [])
movies['keywords']=movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x] if x is not None else [])
movies['cast']=movies['cast'].apply(lambda x:[i.replace(" ","") for i in x] if x is not None else [])
movies['crew']=movies['crew'].apply(lambda x:[i.replace(" ","") for i in x] if x is not None else [])
```

```python
movies.head()
```

```python
movies['tags']=movies['overview']+movies['genres']+movies['keywords']+movies['cast']+movies['crew']
```

```python
newdf=movies[['id','title','tags']]
```

```python
newdf.head()
```

now we have convert to list into string in tag column

```python
newdf['tags']=newdf['tags'].apply(lambda x:" ".join(x))
```

for suggestion of similar movie we apply bag of words(vectorization) technique

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words='english')
# this function return matrix and we have to convert that into arry
vector=cv.fit_transform(newdf['tags']).toarray()
```

```python
vector
```

```python
all_features = cv.get_feature_names_out()
display(list(all_features))
```

stemming is the process in which we combine similar meaning words in single word['love','loves']=['love']

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
cv=PorterStemmer()
```

```python
def stem(text):
  y=[]
  for i in text.split():
    y.append(cv.stem(i))
  return " ".join(y)
```

```python
newdf['tags']=newdf['tags'].apply(stem)
```

we have to find the distance of each vector with each
 vector, we calculated cosin distance(distance btw angles)

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
cosine_similarity(vector)
```

```python
cosine_similarity(vector).shape
```

```python
similarity=cosine_similarity(vector)
```

shape should be in equal numbers because total number of vector for each row are equal.


```python
similarity[0]
```

Now we have to match the movie distance from other movie

```python
sorted(list(enumerate(similarity[0])),reverse=True,key=lambda x:x[1])[1:6]
```

```python
def recommend(movie):
  #index
  movie_index =newdf[newdf['title']== movie].index[0]#masking
  distances= similarity[movie_index]
  movies_list=sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]
  for i in movies_list:
    print(newdf.iloc[i[0]].title)
```

```python
recommend('Gandhi')
```
